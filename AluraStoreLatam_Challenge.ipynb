### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
facturacion = []

suma_1 = sum(tienda.Precio)
suma_2 = sum(tienda2.Precio)
suma_3 = sum(tienda3.Precio)
suma_4 = sum(tienda4.Precio)
#print(f'Venta tienda_4 : {int(suma_4):,}')
#print(f"{numero:,.2f}")  # Resultado: 1,234,567.89

facturacion.append(['Tienda 1',suma_1])
facturacion.append(['Tienda 2',suma_2])
facturacion.append(['Tienda 3',suma_3])
facturacion.append(['Tienda 4',suma_4])

print(facturacion)

mayor = max(facturacion, key=lambda x: x[1])
print(f"{mayor[0]} con Mayor facturacion {mayor[1]:,.0f}")

menor = min(facturacion, key=lambda x: x[1])
print(f"{menor[0]} con menor facturacion {menor[1]:,.0f}")

#print(facturacion)
#print(mayor)
#print(menor)

from matplotlib import pyplot as plt
x = []
y = []

tda = [item[0] for item in facturacion]
y = [item[1] for item in facturacion]

plt.bar(x= tda, height=y)
plt.ylabel('Ventas por Tienda')
plt.show()

# Crear gráfico circular
xxx = plt.pie(y, labels=tda, autopct='%1.1f%%', startangle=90)
yyy = plt.title('Ventas x Tiendas')


```

# 2. Ventas por categoría

```python
# Combinar todos los DataFrames
tiendas = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)



# Calcular monto vendido (asumiendo 1 unidad por fila)
tiendas['Monto Vendido'] = tiendas['Precio']  # Si hay cantidad, sería tiendas['Precio'] * tiendas['Cantidad']

# Agrupar por Categoría del Producto y sumar
resumen = tiendas.groupby('Categoría del Producto')['Monto Vendido'].sum().reset_index()

# Ordenar de mayor a menor (opcional)
resumen = resumen.sort_values(by='Monto Vendido', ascending=False)

print(resumen)

plt.figure(figsize=(10, 6))
plt.bar(resumen['Categoría del Producto'], resumen['Monto Vendido'], color='skyblue')
plt.xlabel('Categoría del Producto')
plt.ylabel('Monto Vendido')
plt.title('Monto Vendido por Categoría del Producto')
plt.xticks(rotation=45, ha='right')  # Rotar etiquetas si son largas
plt.tight_layout()
plt.show()
print("-" * 100)
print("")


# venta por tienda/categoria
resumen1 = tienda.groupby("Categoría del Producto")["Precio"].sum().reset_index()
resumen1.columns = ["categoría", "monto_total"]
resumen1["tienda"] = "Tienda1"
resumen1 = resumen1.sort_values(by="monto_total", ascending=False)

print("\n")
print('Ventas Tienda 1\n')

print(resumen1)
plt.figure(figsize=(10, 6))
plt.bar(resumen1['categoría'], resumen1['monto_total'], color='skyblue')
plt.xlabel('Categoría del Producto')
plt.ylabel('Monto Vendido')
plt.title('Monto Vendido por Categoría del Producto, Tienda 1')
plt.xticks(rotation=45, ha='right')  # Rotar etiquetas si son largas
plt.tight_layout()
plt.show()
print("-" * 100)

print("\n")
print('Ventas Tienda 2\n')

print(resumen2)
plt.figure(figsize=(10, 6))
plt.bar(resumen2['categoría'], resumen2['monto_total'], color='skyblue')
plt.xlabel('Categoría del Producto')
plt.ylabel('Monto Vendido')
plt.title('Monto Vendido por Categoría del Producto, Tienda 2')
plt.xticks(rotation=45, ha='right')  # Rotar etiquetas si son largas
plt.tight_layout()
plt.show()
print("-" * 100)

print("\n")
print('Ventas Tienda 3\n')

print(resumen3)
plt.figure(figsize=(10, 6))
plt.bar(resumen3['categoría'], resumen3['monto_total'], color='skyblue')
plt.xlabel('Categoría del Producto')
plt.ylabel('Monto Vendido')
plt.title('Monto Vendido por Categoría del Producto, Tienda 3')
plt.xticks(rotation=45, ha='right')  # Rotar etiquetas si son largas
plt.tight_layout()
plt.show()
print("-" * 100)

print("\n")
print('Ventas Tienda 4\n')
print(resumen4)

plt.figure(figsize=(10, 6))
plt.bar(resumen4['categoría'], resumen4['monto_total'], color='skyblue')
plt.xlabel('Categoría del Producto')
plt.ylabel('Monto Vendido')
plt.title('Monto Vendido por Categoría del Producto, Tienda 4')
plt.xticks(rotation=45, ha='right')  # Rotar etiquetas si son largas
plt.tight_layout()
plt.show()

```

# 3. Calificación promedio de la tienda


```python
promedio_tienda1 = tienda["Calificación"].mean()
print("Promedio de calificación Tienda 1:", promedio_tienda1)

promedio_tienda2 = tienda2["Calificación"].mean()
print("Promedio de calificación Tienda 2:", promedio_tienda2)

promedio_tienda3 = tienda3["Calificación"].mean()
print("Promedio de calificación Tienda 3:", promedio_tienda3)

promedio_tienda4 = tienda4["Calificación"].mean()
print("Promedio de calificación Tienda 1:", promedio_tienda4)
```

# 4. Productos más y menos vendidos

```python

# calcula producto con mayor y menor venta Tienda 1
conteo_productos = tienda["Producto"].value_counts()

producto_mas_vendido = conteo_productos.idxmax()
cantidad_mas_vendido = conteo_productos.max()
producto_menos_vendido = conteo_productos.idxmin()
cantidad_menos_vendido = conteo_productos.min()

ventas_mas_vendido = tienda[tienda["Producto"] == producto_mas_vendido]
ventas_menos_vendido = tienda[tienda["Producto"] == producto_menos_vendido]
precio_promedio_mas_vendido = ventas_mas_vendido["Precio"].mean()
precio_promedio_menos_vendido = ventas_menos_vendido["Precio"].mean()

print("-"*100)
print(f"Producto más vendido TIENDA 1 : {producto_mas_vendido} ({cantidad_mas_vendido} ventas) - Precio promedio: ${precio_promedio_mas_vendido:.2f}")
print(f"Producto menos vendido TIENDA 1: {producto_menos_vendido} ({cantidad_menos_vendido} ventas) - Precio promedio: ${precio_promedio_menos_vendido:.2f}")
print('\n')
print("-"*100)



# calcula producto con mayor y menor venta Tienda 2
conteo_productos = tienda2["Producto"].value_counts()

producto_mas_vendido2   = conteo_productos.idxmax()
cantidad_mas_vendido2   = conteo_productos.max()
producto_menos_vendido2 = conteo_productos.idxmin()
cantidad_menos_vendido2 = conteo_productos.min()

ventas_mas_vendido2 = tienda2[tienda["Producto"] == producto_mas_vendido]
ventas_menos_vendido2 = tienda2[tienda["Producto"] == producto_menos_vendido]
precio_promedio_mas_vendido2 = ventas_mas_vendido2["Precio"].mean()
precio_promedio_menos_vendido2 = ventas_menos_vendido2["Precio"].mean()

print(f"Producto más vendido TIENDA 2: {producto_mas_vendido2} ({cantidad_mas_vendido2} ventas) - Precio promedio: ${precio_promedio_mas_vendido2:.2f}")
print(f"Producto menos vendido TIENDA 2: {producto_menos_vendido2} ({cantidad_menos_vendido2} ventas) - Precio promedio: ${precio_promedio_menos_vendido2:.2f}")
print('\n')
print("-"*100)

# calcula producto con mayor y menor venta Tienda 3
conteo_productos = tienda3["Producto"].value_counts()

producto_mas_vendido3   = conteo_productos.idxmax()
cantidad_mas_vendido3   = conteo_productos.max()
producto_menos_vendido3 = conteo_productos.idxmin()
cantidad_menos_vendido3 = conteo_productos.min()

ventas_mas_vendido3 = tienda3[tienda["Producto"] == producto_mas_vendido]
ventas_menos_vendido3 = tienda3[tienda["Producto"] == producto_menos_vendido]
precio_promedio_mas_vendido3 = ventas_mas_vendido3["Precio"].mean()
precio_promedio_menos_vendido3 = ventas_menos_vendido3["Precio"].mean()
print(f"Producto más vendido TIENDA 3: {producto_mas_vendido3} ({cantidad_mas_vendido3} ventas) - Precio promedio: ${precio_promedio_mas_vendido3:.2f}")
print(f"Producto menos vendido TIENDA 3: {producto_menos_vendido3} ({cantidad_menos_vendido3} ventas) - Precio promedio: ${precio_promedio_menos_vendido3:.2f}")
print('\n')
print("-"*100)

# calcula producto con mayor y menor venta Tienda 4
conteo_productos = tienda4["Producto"].value_counts()

producto_mas_vendido4   = conteo_productos.idxmax()
cantidad_mas_vendido4   = conteo_productos.max()
producto_menos_vendido4 = conteo_productos.idxmin()
cantidad_menos_vendido4 = conteo_productos.min()

ventas_mas_vendido4 = tienda4[tienda["Producto"] == producto_mas_vendido]
ventas_menos_vendido4 = tienda4[tienda["Producto"] == producto_menos_vendido]
precio_promedio_mas_vendido4 = ventas_mas_vendido3["Precio"].mean()
precio_promedio_menos_vendido4 = ventas_menos_vendido3["Precio"].mean()
print(f"Producto más vendido TIENDA 4: {producto_mas_vendido4} ({cantidad_mas_vendido4} ventas) - Precio promedio: ${precio_promedio_mas_vendido4:.2f}")
print(f"Producto menos vendido TIENDA 4: {producto_menos_vendido4} ({cantidad_menos_vendido4} ventas) - Precio promedio: ${precio_promedio_menos_vendido4:.2f}")
print('\n')
print("-"*100)
```
